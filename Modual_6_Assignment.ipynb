<h3><u>This cell aims to introduce the reader to the basic arithmetic expressions we all know and love</u></h3>

<b>Addition:</b> the process of calculating the total of two or more numbers: 1+1=2<br>
<b>Subtraction:</b> the process of removing a value from another: 1-1=0<br>
<b>Multiplication:</b> the process of calculating the product of two or more numbers: 1x1=1<br>
<b>Division:</b> the process of breaking a number up into equal parts, and finding out how many equal parts can be made: 1/1=1

```python
from functools import reduce
from operator import mul


def multiply_numbers(numbers_to_use: list = [], multiply=False, adding=False):
    results = {}
    if multiply is True:
        default = 1
        results['multiply_result'] = reduce(mul, numbers_to_use)
    if adding is True:
        results['adding_result'] = sum(numbers_to_use)
    return results


print(multiply_numbers([1, 2], True, True))
```

```python
def minutes_to_hours(minute_list: list):
    x = [minute/60 if minute > 0 else None for minute in minute_list]
    return x

minutes_to_hours([60, 0, 20])
```
